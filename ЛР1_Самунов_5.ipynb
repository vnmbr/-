### 4. Проверка дубликатов

#### Проверка явных дубликатов

```python
df.duplicated().sum()
```

Явные дупликаты отсутствуют

#### Проверка неявных дубликатов

```python
print('region:', df['region'].unique())
print('device:', df['device'].unique())
print('channel:', df['channel'].unique())
```

после проверки можно заметить что явно одинаковые вещи разными пользователями могут быть записаны по разному, например: 'United States' и 'USA', 'Mac' и 'MAC', 'iPhone' и 'IPHONE'. Поэтому необходимо привести эти значения к единому виду и провести провекру на дубликаты ещё раз.

```python
df['region'] = df['region'].replace('United States', 'USA')
df['device'] = df['device'].replace({
    'MAC': 'Mac',
    'IPHONE': 'iPhone'
})

print('region:', df['region'].unique())
print('device:', df['device'].unique())
```

Теперь неоднозначных значений нет

```python
df.duplicated().sum()
```

После приведения неоднознеачных значений к единому виду явных дубликатов не появилось

### 5. Провека типов данных

```python
print((df['click_count'] % 1 == 0).all())
print((df['buy_count'] % 1 == 0).all())
print((df['time_session'] % 1 == 0).all())
```

```python
#Три столбца с форматом данных float64, но в них по логике должны быть только целые числа. После проверки видно что числа действительно целыее, так что преобразуем их в формат int64.
df['click_count'] = df['click_count'].astype('int64')
df['buy_count'] = df['buy_count'].astype('int64')
df['time_session'] = df['time_session'].astype('int64')
```

Столбцы session_start и session_end содержат дату и время, поэтому их надо преобразовать из типа object в datetime.
Остальные столбцы находятся в подходящих им форматах.

```python
df['session_start'] = pd.to_datetime(df['session_start'])
df['session_end'] = pd.to_datetime(df['session_end'])
```

### 6. Группировка данных

#### Задание 1

*`Группировка - device и количество рекламных источников каждого типа (channel)`*

```python
# выполните группировку согласно варианту
df.groupby(['device', 'channel']).size()
```

По результатам группировки видно, что для всех типов устройств наибольшее количество записей приходится на рекламный канал organic. Наименьшее количество записей - на канал MediaTornado. Среди устройств наибольшее общее количество записей связано с iPhone, наименьшее - с PC.

#### Задание 2

*`Группировка - region и количество устройств (device). Создать датафрейм. Переименовать столбец с количеством в “сount”. Отсортировать по убыванию столбца “count”.`*

```python
# выполните группировку согласно варианту
ndf = df.groupby('region')['device'].count().reset_index()
ndf = ndf.rename(columns={'device': 'count'})
ndf.sort_values(by='count', ascending = False)
```

Видно что из США записей коллосально больше чем из России. Из работы выше можно постановить что все данные были отобраны - всего регионов действительно два, а в сумме они дают все 952 записи.

#### Задание 3

*`Сводная таблица (pivot_table) - уникальное количество пользователей для каждого устройства (device). Отсортировать по убыванию количества.`*

```python
# выполните сводную таблицу согласно варианту
df3 = df.pivot_table(index='device', values='user_id', aggfunc='nunique')
df3.sort_values(by='user_id', ascending = False)
```

из результатов сводной таблицы видно количетсво пользователей для каждого устройства. Наибольшее число пользователей приходится на iPhone, наименьшее - на PC.

#### Задание 4

*`Сводная таблица (pivot_table) - уникальное количество пользователей для каждого устройства (device) - столбцы и регионастроки. Отсортировать по возрастанию столбца region.`*

```python
# выполните сводную таблицу согласно варианту
df4 = df.pivot_table(index='region', columns='device', values='user_id', aggfunc='nunique')
df4.sort_index(ascending=True)
```

Из результатов сводной таблицы видно, что в обоих регионах наибольшее количество пользователей приходится на iPhone. В России iPhone используют 23 пользователя, Mac - 12, PC - 6, Android - 11. В США количество пользователей значительно выше, что подтверждает результаты задания 2: iPhone - 368, Mac - 220, PC - 97, Android - 167. Таким образом, iPhone является самым распространённым устройством среди пользователей в обоих регионах, а PC - наименее распространённым.
