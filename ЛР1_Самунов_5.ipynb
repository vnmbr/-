Как можно понять изначально был выбран неверный формат перевода данных. В лабораторной работе был использован формат date-time по умолчанию и из строчек выше видно что он явно не подходит, месяц и день перепутаны местами. для этого преобразуем данные правильно.

```python
for column in ['session_start', 'session_end']:
    as_string = df[column].dt.strftime('%Y-%m-%d %H:%M:%S')
    df[column] = pd.to_datetime(as_string, format='%Y-%d-%m %H:%M:%S')

df['session_dur'] = df['session_end'] - df['session_start']

df.loc[df['session_dur'] > pd.Timedelta(hours=5)]
```

Выбрать только те записи, средняя длительность на которых лежит в определенном диапазоне (нижнюю и верхнюю границы выбрать самостоятельно) + устройство = Android и PC+ 2 канала с минимальным количеством записей. Для отфильтрованного набора данных создать сводную таблицу: канал, устройство и средняя длительность сессии.`**

```python
# код выполнения задания 20
lowest_channels = df['channel'].value_counts().nsmallest(2).index

df20 = df[
    df['session_dur'].between(
        pd.Timedelta(minutes=10), pd.Timedelta(minutes=45)
    )
    & (df['device'].isin(['Android', 'PC']))
    & (df['channel'].isin(lowest_channels))
]

print('Два канала с минимальным количеством записей:', list(lowest_channels))
df20.pivot_table(
    index='channel',
    columns='device',
    values='session_dur',
    aggfunc='mean'
)
```
